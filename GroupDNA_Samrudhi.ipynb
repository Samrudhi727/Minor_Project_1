## FEATURE 1: CHAT PARSER

```python
with open("hostel_bois.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

print("Total lines in file:", len(lines))
```

```python
message_list = []

system_messages = 0
media_messages = 0
deleted_messages = 0

for row in lines:

    row = row.strip()

    if row == "":
        continue

    timestamp, content = row.split(" - ", 1)

    try:

        user, message_text = content.split(": ", 1)

        if message_text == "<Media omitted>":
            media_messages += 1

        elif message_text == "This message was deleted":
            deleted_messages += 1

        message_data = {
            "timestamp": timestamp,
            "sender": user,
            "text": message_text
        }

        message_list.append(message_data)

    except:
        system_messages += 1

print("Parsed Messages :", len(message_list))
print("System Messages :", system_messages)
print("Media Omitted   :", media_messages)
print("Deleted Messages:", deleted_messages)
```

```python
# Last 5 messages of chat

for i in range(-5, 0):
    print(lines[i])
```

## Feature 2: Group Overview

```python
msg_counter = {}
for row in message_list:
    who = row["sender"]
    if who not in msg_counter:
        msg_counter[who] = 0
    msg_counter[who] += 1

date_list = []
for row in message_list:
    dt = datetime.strptime(row["timestamp"], "%d/%m/%y, %H:%M")
    date_list.append(dt.date())

start_day = min(date_list)
end_day = max(date_list)
num_days = (end_day - start_day).days + 1
total_msgs = len(message_list)

print()
print("=" * 60)
print(" GROUP OVERVIEW")
print("=" * 60)
print(" Period :", start_day.strftime("%d %B %Y"), "to", end_day.strftime("%d %B %Y"), f"({num_days} days)")
print(" Total messages :", total_msgs)
print(" Participants   :", len(msg_counter))
print()
```

```python
print("GROUP PARTICIPATION")

for person, count in sorted(msg_counter.items(), key=lambda x: x[1], reverse=True):

    percentage = count / len(message_list) * 100

    print(person, ":", count, "messages", f"({percentage:.2f}%)")
```

## Feature 3: Most Active Day and Hour

```python
day_count = {}

for msg in message_list:

    date = msg["timestamp"].split(",")[0]

    if date in day_count:
        day_count[date] += 1
    else:
        day_count[date] = 1

print(day_count)
```

```python
hour_count = {}

for msg in message_list:

    hour = msg["timestamp"].split(",")[1].strip().split(":")[0]

    if hour in hour_count:
        hour_count[hour] += 1
    else:
        hour_count[hour] = 1

print(hour_count)
```

```python
busiest_day = max(day_count, key=day_count.get)

print("BUSIEST DAY")
print("Busiest Day :", busiest_day)
print("Messages    :", day_count[busiest_day])

busiest_hour = max(hour_count, key=hour_count.get)

print("\nBUSIEST HOUR")
print("Busiest Hour :", busiest_hour)
print("Messages     :", hour_count[busiest_hour])
```

## Feature 4: Activity Heatmap (NumPy)

```python
import numpy as np
from datetime import datetime

users = sorted(list(msg_counter.keys()))

user_index = {}

for i, user in enumerate(users):
    user_index[user] = i

heatmap = np.zeros((len(users), 24), dtype=int)

for msg in message_list:

    sender = msg["sender"]

    dt = datetime.strptime(
        msg["timestamp"],
        "%d/%m/%y, %H:%M"
    )

    hour = dt.hour

    heatmap[user_index[sender]][hour] += 1

print("Activity analysis completed")

print(f"\n{len(users)} x 24 ACTIVITY MATRIX")
print(heatmap)
```

```python
# AI assistance was used for guidance in implementing this activity heatmap.
# █ High | ▒ Medium | ░ Low | . Very Low

print("\nACTIVITY HEATMAP")

show_hours = [0, 3, 6, 9, 12, 15, 18, 21]

header = "        "

for hour in show_hours:
    header += f"{hour:02d} "

print(header)

for user in users:

    row = heatmap[user_index[user]]
    maximum = row.max()

    line = f"{user:<8}"

    for hour in show_hours:

        value = row[hour]

        if maximum == 0:
            block = "."
        else:
            ratio = value / maximum

            if ratio >= 0.75:
                block = "█"
            elif ratio >= 0.50:
                block = "▒"
            elif ratio >= 0.25:
                block = "░"
            else:
                block = "."

        line += f"{block}  "

    print(line)
```

## Feature 6: Response Speed & Silent Streaks

```python
response_times = {}

for sender in msg_counter:

    response_times[sender] = []

for i in range(1, len(message_list)):

    current = message_list[i]

    previous = message_list[i - 1]

    current_sender = current["sender"]

    previous_sender = previous["sender"]

    if current_sender != previous_sender:

        current_time = datetime.strptime(
            current["timestamp"],
            "%d/%m/%y, %H:%M"
        )

        previous_time = datetime.strptime(
            previous["timestamp"],
            "%d/%m/%y, %H:%M"
        )

        gap = current_time - previous_time

        minutes = gap.total_seconds() / 60

        response_times[current_sender].append(minutes)
```

```python
average_response = {}

for sender in response_times:

    times = response_times[sender]

    if len(times) == 0:

        average_response[sender] = 0

    else:

        total = 0

        for time in times:

            total += time

        average = total / len(times)

        average_response[sender] = average
```

```python
print("Average response times:\n")

for sender in average_response:

    print(
        sender,
        ":",
        round(average_response[sender], 2),
        "minutes"
    )
```

```python
longest_streak = {}

for sender in msg_counter:

    longest_streak[sender] = 0

for sender in msg_counter:

    dates = []

    for message in message_list:

        if message["sender"] == sender:

            dt = datetime.strptime(
                message["timestamp"],
                "%d/%m/%y, %H:%M"
            )

            current_date = dt.date()

            if current_date not in dates:

                dates.append(current_date)

    max_gap = 0

    for i in range(1, len(dates)):

        gap = (dates[i] - dates[i - 1]).days - 1

        if gap > max_gap:

            max_gap = gap

    longest_streak[sender] = max_gap

print("Longest silent streak:\n")

for sender in longest_streak:

    print(sender, ":", longest_streak[sender], "days")
```

## Feature 7: Personality Archetype Detection

```python
# Spammer

spammer_count = {}

for msg in message_list:

    sender = msg["sender"]

    if sender in spammer_count:
        spammer_count[sender] += 1
    else:
        spammer_count[sender] = 1

spammer = max(spammer_count, key=spammer_count.get)

print("Group Spammer :", spammer)
```

```python
# Group Mom

mom_words = [
    "take care", "eat", "sleep", "study", "rest", "drink water"
]

mom_count = {}

for msg in message_list:

    sender = msg["sender"]
    text = msg["text"].lower()

    for word in mom_words:
        if word in text:
            mom_count[sender] = mom_count.get(sender, 0) + 1

group_mom = max(mom_count, key=mom_count.get)

print("Group Mom :", group_mom)
```

```python
# Night Owl

night_count = {}

for msg in message_list:

    sender = msg["sender"]

    hour = int(msg["timestamp"].split(",")[1].strip().split(":")[0])

    if hour >= 0 and hour < 5:
        night_count[sender] = night_count.get(sender, 0) + 1

night_owl = max(night_count, key=night_count.get)

print("Night Owl :", night_owl)
```

```python
# Storyteller

story_count = {}

for msg in message_list:

    sender = msg["sender"]
    text = msg["text"]

    word_count = len(text.split())

    if word_count >= 30:
        story_count[sender] = story_count.get(sender, 0) + 1

storyteller = max(story_count, key=story_count.get)

print("Storyteller :", storyteller)
```

```python
# Drama Queen

drama_words = [
    "omg", "seriously", "what", "why", "no way",
    "unbelievable", "wtf", "damn", "ugh"
]

drama_count = {}

for msg in message_list:

    sender = msg["sender"]
    text = msg["text"].lower()

    for word in drama_words:
        if word in text:
            drama_count[sender] = drama_count.get(sender, 0) + 1

drama_queen = max(drama_count, key=drama_count.get)

print("Drama Queen :", drama_queen)
```

```python
# Ghost

ghost_count = {}

for msg in message_list:

    sender = msg["sender"]

    if sender not in ghost_count:
        ghost_count[sender] = 0

    ghost_count[sender] += 1

ghost = min(ghost_count, key=ghost_count.get)

print("Group Ghost :", ghost)
```

```python
# Comedian

funny_words = [
    "lol", "lmao", "haha", "hahaha", "😂", "🤣",
    "rofl", "hehe"
]

funny_count = {}

for msg in message_list:

    sender = msg["sender"]
    text = msg["text"].lower()

    for word in funny_words:
        if word in text:
            funny_count[sender] = funny_count.get(sender, 0) + 1

comedian = max(funny_count, key=funny_count.get)

print("Group Comedian :", comedian)
```
